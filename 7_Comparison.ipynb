```python
import matplotlib.pyplot as plt
import numpy as np

targets = ['Distillate Purity', 'Bottom Purity', 'Condenser Duty', 'Reboiler Duty']
r2_extra_trees = [0.9515, 0.9475, 0.9715, 0.9419]
r2_svr         = [0.9133, 0.9100, 0.9881, 0.9897]
r2_ann         = [0.9938, 0.9541, 0.9881, 0.9916]

x = np.arange(len(targets))
w = 0.25

plt.figure(figsize=(10, 6))
plt.bar(x - w, r2_extra_trees, w, label='ExtraTrees / XGBoost', color='#4c72b0')
plt.bar(x, r2_svr, w, label='Support Vector Regressor (SVR)', color='#55a868')
plt.bar(x + w, r2_ann, w, label='Artificial Neural Network (ANN)', color='#c44e52')

plt.ylabel('$R^2$ Score (Higher is Better)', fontweight='bold')
plt.title('Master Model Performance Comparison Across All 4 Targets', fontweight='bold', fontsize=14)
plt.xticks(x, targets, fontweight='bold')
plt.ylim([0.85, 1.02])
plt.legend(loc='lower right')
plt.tight_layout()
plt.savefig("master_model_comparison.png", dpi=300)
plt.show()
```
