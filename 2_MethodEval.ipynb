```python
# Adding references for out pythonnet 
clr.AddReference("DWSIM.Automation")
clr.AddReference("DWSIM.Interfaces")
```

```python
#loading in automation 
from DWSIM.Automation import Automation3
automation = Automation3()
```

```python
#loading iin flowsheet 
flowsheet = automation.LoadFlowsheet(
    r"C:\Users\Admin\AppData\Local\DWSIM\bdc.dwxmz" # addr. of a dwsim project with a basic setup of binary dis. column 
)
print("Flowsheet Loaded !")
```

```python
#accesssing feed stream, column , up stream and bottom stream by initiating them as follow
feed = flowsheet.GetFlowsheetSimulationObject("Feed") #feed stream was implicitly names as "1" i named it to feed in dwsim 
col = flowsheet.GetFlowsheetSimulationObject("DCOL-1") #column initiation 
dis= flowsheet.GetFlowsheetSimulationObject("disStream") #upstream initiation
btm = flowsheet.GetFlowsheetSimulationObject("bottomStream") #Bottom stream initiation 
```

```python
#checking out all methods we get with simulation objects to get a idea of what might be useful
dir(feed)
```

```python
#finding out what arguments the intrested method (ex. GetProperties , Getproperties2 ) takes
t = feed.GetType()
for m in t.GetMethods():
    if m.Name == "GetProperties":
        print(m)
```

```python
#finding out what arguments GetPropertyValue() takes
t = feed.GetType()
for m in t.GetMethods():
    if m.Name == "GetPropertyValue":
        print(m)
```

```python
#lets define a variable names units so that we can pass it as an argument to GetPropertyValue() method 
units = flowsheet.FlowsheetOptions.SelectedUnitSystem
```

```python
# so its Enums and PropertyType in it , (DWSIM have all its properties lables as ALL,RW,RO,WR)
from DWSIM.Interfaces.Enums import PropertyType
```

```python
#checking out labels of properties with property type ALL 
mode = PropertyType.ALL
print("========",mode,"==========")
for obj in [feed,col,dis,btm]:
    print("=========",obj,"============")
    for prop in obj.GetProperties(mode):
        try:
            print(prop, "=", obj.GetPropertyValue(prop, units))
        except Exception as e:
            print(prop, e)
     #this gives us lable names 
        
```
